```python
import numpy as np

def vae_encoder(x: np.ndarray, W_mu: np.ndarray, b_mu: np.ndarray, W_logvar: np.ndarray, b_logvar: np.ndarray) -> dict:
    """
    Returns: dict with 'mu' and 'log_var' as np.ndarrays of shape (batch, latent_dim)
    """
    mu = x @ W_mu + b_mu
    log_var = x@W_logvar + b_logvar
    return {'mu':mu,"log_var":log_var}
```

```python
x = np.array([[0.3929,-0.4277,-0.5463],[0.1026,0.4389,-0.1538]])
W_mu = np.array([[0.4808,0.1848],[-0.0191,-0.1079],[-0.1568,0.229]])
b_mu = np.array([-0.0369,-0.2642])
W_logvar = np.array([[-0.102,0.238],[-0.3175,-0.3245],[0.0316,0.0318]])
b_logvar = np.array([0.0806,0.2097])

vae_encoder(x,W_mu,b_mu,W_logvar,b_logvar)
```
