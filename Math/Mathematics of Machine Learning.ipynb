## The Basis

The vector spaces contain infinite sets of vectors, but we can represent all the vectors in a vector space using a finite set of vectors called a basis. A basis is a set of linearly independent vectors that span the entire vector space.

We can split the the vectors $x={x_1,x_2, \ldots, x_n}$ to a basis set $E={e_1, e_2, \ldots, e_m}$ and coefficients $c={c_1, c_2, \ldots, c_m}$ such that:

$$x = c_1 e_1 + c_2 e_2 + \ldots + c_m e_m$$

$$x = \sum_{i=1}^m c_i e_i$$


## Linear combinations and independence

### Definition 3. (Linear dependence and independence)

Let V be a vector space and $S = {v_1,\ldots,v_n}$ be a subset of its vectors. S is said to be linearly dependent if it only contains the zero vector, or there is a nonzero $v_k$ that can be expressed as a linear combination of the other vectors $v_1,\ldots,v_{k-1},v_{k+1},\ldots,v_n$.

S is said to be linearly independent if it is not linearly dependent.


#### Example: Linear Dependence and Independence

##### Example 1: Linearly Dependent Vectors

Let $V = \mathbb{R}^2$ and $S = \{v_1, v_2, v_3\}$ where:

- $v_1 = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$
- $v_2 = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$
- $v_3 = \begin{pmatrix} 1 \\ 1 \end{pmatrix}$

S is linearly dependent because $v_3$ can be expressed as a linear combination of $v_1$ and $v_2$:

$$v_3 = 1 \cdot v_1 + 1 \cdot v_2 = \begin{pmatrix} 1 \\ 1 \end{pmatrix}$$

##### Example 2: Linearly Independent Vectors

Let $V = \mathbb{R}^2$ and $S = \{v_1, v_2\}$ where:

- $v_1 = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$
- $v_2 = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$

S is linearly independent because neither vector can be expressed as a linear combination of the other. The only solution to $c_1 v_1 + c_2 v_2 = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$ is $c_1 = 0$ and $c_2 = 0$.


### Theorem 2.

Let V be a vector space and $S = {v_1,\ldots,v_n}$ be a subset of its vectors.

(a) S is linearly dependent if and only if the null vector 0 can be obtained as a nontrivial linear combination.

(b) S is linearly independent if and only if whenever $0 = \sum_{i=1}^n x_i v_i$, all coefficients $x_i$ are zero.


#### Examples: Theorem 2

##### Example 1: Linearly Dependent Vectors (Part a)

Let $V = \mathbb{R}^3$ and $S = \{v_1, v_2, v_3\}$ where:

- $v_1 = \begin{pmatrix} 1 \\ 0 \\ 0 \end{pmatrix}$
- $v_2 = \begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix}$
- $v_3 = \begin{pmatrix} 1 \\ 1 \\ 0 \end{pmatrix}$

According to Theorem 2(a), S is linearly dependent because we can obtain the null vector as a nontrivial linear combination:

$$1 \cdot v_1 + 1 \cdot v_2 - 1 \cdot v_3 = \begin{pmatrix} 1 \\ 0 \\ 0 \end{pmatrix} + \begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix} - \begin{pmatrix} 1 \\ 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \\ 0 \end{pmatrix}$$

The coefficients (1, 1, -1) are not all zero, making this a nontrivial linear combination.

##### Example 2: Linearly Independent Vectors (Part b)

Let $V = \mathbb{R}^2$ and $S = \{v_1, v_2\}$ where:

- $v_1 = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$
- $v_2 = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$

According to Theorem 2(b), S is linearly independent because the only way to express the null vector as a linear combination is:

$$x_1 v_1 + x_2 v_2 = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

implies:

$$x_1 \begin{pmatrix} 1 \\ 0 \end{pmatrix} + x_2 \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This forces $x_1 = 0$ and $x_2 = 0$, so all coefficients must be zero.


## Spans

The span of a set of vectors is the collection of all possible linear combinations of those vectors. If we have a set of vectors $S = {v_1, v_2, \ldots, v_n}$ in a vector space V, the span of S, denoted as span(S), is defined as:

$$\text{span}(S) = \left\{ \sum_{i=1}^n c_i v_i \mid c_i \in \mathbb{R} \right\}$$



### Proposition 1. Let V be a vector space and S,S1,S2 ⊆V be subsets of its vectors.

(a) If $S1 ⊆ S2$, then $\text{span}(S1) ⊆ \text{span}(S2)$.

(b) $\text{span}(\text{span}(S)) = \text{span}(S)$.
