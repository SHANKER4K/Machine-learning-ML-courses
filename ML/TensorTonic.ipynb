```python
import numpy as np

def _sigmoid(z):
    """Numerically stable sigmoid implementation."""
    return np.where(z >= 0, 1/(1+np.exp(-z)), np.exp(z)/(1+np.exp(z)))

def train_logistic_regression(X, y, lr=0.1, steps=1000):
    """
    Train logistic regression via gradient descent.
    Return (w, b).
    """
    X = np.array(X)
    y = np.array(y)
    n_samples, n_features = X.shape
    
    w = np.zeros(n_features)
    b = 0.0
    
    for i in range(steps):
        p = _sigmoid(X @ w + b)
        dw = (X.T @ (p - y)) / n_samples
        db = np.mean(p - y)
        
        w -= lr * dw
        b -= lr * db
    
    return w, b
    
```

```python
def pad_sequences(seqs, pad_value=0, max_len=None):
    """
    Returns: np.ndarray of shape (N, L) where:
      N = len(seqs)
      L = max_len if provided else max(len(seq) for seq in seqs) or 0
    """
    L = max_len
    if max_len == None:

        L = max(len(seq) for seq in seqs)
    arr = np.full((len(seqs),L),pad_value)
    for i in range(len(arr)):
        s = seqs[i]
        for k in range(min(L,len(s))):
            arr[i,k] = s[k]
    return arr
        
```

```python
def gradient_descent_quadratic(a, b, c, x0, lr, steps):
    """
    Return final x after 'steps' iterations.
    """
    def f(x):
        return a*x**2+b*x+c
    def df(x):
        return 2*a*x+b
    for i in range(steps):
        x0 -= lr*df(x0)
    return x0
```

```python
import numpy as np

def adam_step(param, grad, m, v, t, lr=1e-3, beta1=0.9, beta2=0.999, eps=1e-8):
    """
    One Adam optimizer update step.
    Return (param_new, m_new, v_new).
    """
    # hyperparams to np arrays
    beta1 = np.array(beta1 ,dtype=np.float64)
    beta2 = np.array(beta2, dtype=np.float64)
    param = np.array(param, dtype=np.float64)
    grad = np.array(grad, dtype=np.float64)
    m = np.array(m, dtype=np.float64)
    v = np.array(v, dtype=np.float64)

    m = beta1 * m + (1 - beta1) * grad
    v = beta2 * v + (1 - beta2) * grad**2
    
    m_new = m/(1-beta1**t)
    v_new = v/(1-beta2**t)
    
    param -= lr * m_new/(np.sqrt(v_new+eps))
    return (param,m,v)
```

```python
import numpy as np

def rmsprop_step(w, g, s, lr=0.001, beta=0.9, eps=1e-8):
    """
    Perform one RMSProp update step.
    """
    w = np.array(w,dtype=np.float64)
    g = np.array(g,dtype=np.float64)
    s = np.array(s,dtype=np.float64)
    
    s = beta * s + (1 - beta) * g**2
    w -= lr/np.sqrt(s+eps) * g
    
    return w,s
```

```python
import numpy as np

def expected_value_discrete(x, p):
    """
    Returns: float expected value
    """
    x = np.array(x)
    p = np.array(p)
    
    if not np.isclose(np.sum(p), 1.0):
        raise ValueError("Probabilities must sum to 1.")
        
    return np.sum(x * p)
```

```python
import numpy as np

def entropy_node(y):
    """
    Compute entropy for a single node using stable logarithms.
    """
    _ ,counts = np.unique(y,return_counts=True)
    p = counts/len(y)
    return np.sum(-p*np.log2(p))
```
