```python
import numpy as np

X = np.random.randn(2,3)
w = np.random.randn(3,2)
b = np.random.randn(2)
```

```python
def perc(X,w,b):
    return X @ w + b
```

```python
z1 = perc(X,w,b)
z1
```

```python
def ReLU(X):
    return np.maximum(0,X)
ReLU(z1)
```

```python
def sigmoid(X):
    return 1/(1+np.exp(-X))
sigmoid(z1)
```
